```python
##Import libraries

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
#plot log returns with time stamps converted to datetime
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
df['Log_Returns'] = np.log(df['Close'] / df['Close'].shift(1))
plt.figure(figsize=(12,6))
plt.plot(df['Timestamp'], df['Log_Returns'], color='orange')
plt.title('Bitcoin Log Returns Over Time')
plt.xlabel('Timestamp')
plt.ylabel('Log Returns')
plt.show()
```

```python
# acf calculation and plot
from statsmodels.graphics.tsaplots import plot_acf
from matplotlib import pyplot as plt
plt.figure(figsize=(10,5))
plot_acf(df['Log_Returns'].dropna(), lags=50)
plt.title('Autocorrelation of Bitcoin Log Returns')
plt.xlabel('Lags')
```
